```python
# Dataset
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

# 1. Number of samples
num_samples = len(study_hours)
print("Number of samples:", num_samples)

# 2. Average study hours
average_hours = sum(study_hours) / num_samples
print("Average study hours:", average_hours)

# 3. Average marks
average_marks = sum(marks) / num_samples
print("Average mark:", average_marks)

# 4. Highest and lowest marks
highest_mark = max(marks)
lowest_mark = min(marks)

print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)

# 5. Display each study-hours/mark pair
print("\nStudy Hours and Marks:")
for hours, mark in zip(study_hours, marks):
    print("Study hours:", hours, "- Mark:", mark)

# 6. Display Pass or Fail for each sample
print("\nPass/Fail Results:")
for hours, mark in zip(study_hours, marks):
    if mark >= 60:
        result = "Pass"
    else:
        result = "Fail"

    print("Study hours:", hours, "- Mark:", mark, "-", result)

# 7. Count the number of passes
pass_count = 0

for mark in marks:
    if mark >= 60:
        pass_count += 1

print("\nNumber of students who passed:", pass_count)
```
