## Part 2: Defining the Machine Learning Problem

### Machine Learning Type
Regression

### Target Variable
`median_earnings_4yr_usd`

### Problem Description
The objective is to predict the median earnings of graduates four years after completing a college program. The prediction will be based on available characteristics of the institution, academic program, tuition, admission, student outcomes, debt, and related occupation information.

This is a regression problem because the target variable, `median_earnings_4yr_usd`, is a numerical continuous value measured in US dollars.

## Part 4: Methodology Diagram

The methodology describes the steps that will be followed to build a regression model for predicting 4-year median earnings.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

# Create figure
fig, ax = plt.subplots(figsize=(10, 14))

# Hide axes
ax.set_xlim(0, 10)
ax.set_ylim(0, 14)
ax.axis("off")

# Steps in the methodology
steps = [
    ("1. Dataset Selection",
     "College Major & Career Outcomes Dataset"),

    ("2. Data Loading",
     "Load the CSV dataset using Pandas"),

    ("3. Data Preprocessing",
     "Handle missing values\nSelect relevant features\nPrepare data for modeling"),

    ("4. Train/Test Split",
     "Split the data into training and testing sets"),

    ("5. Model Training",
     "Train a Regression Model\nusing the selected features"),

    ("6. Model Evaluation",
     "Evaluate predictions using\nMAE, RMSE, and R²"),

    ("7. Results",
     "Analyze the model performance\nand predicted earnings")
]

# Vertical positions
y_positions = [12.5, 10.7, 8.7, 6.7, 4.7, 2.7, 0.7]

# Draw boxes
for (title, description), y in zip(steps, y_positions):
    box = FancyBboxPatch(
        (1.5, y - 0.6),
        7,
        1.2,
        boxstyle="round,pad=0.05",
        linewidth=2,
        edgecolor="black",
        facecolor="white"
    )
    ax.add_patch(box)

    ax.text(
        5,
        y + 0.15,
        title,
        ha="center",
        va="center",
        fontsize=13,
        fontweight="bold"
    )

    ax.text(
        5,
        y - 0.25,
        description,
        ha="center",
        va="center",
        fontsize=10
    )

# Draw arrows
for i in range(len(y_positions) - 1):
    ax.annotate(
        "",
        xy=(5, y_positions[i + 1] + 0.6),
        xytext=(5, y_positions[i] - 0.6),
        arrowprops=dict(arrowstyle="->", linewidth=2)
    )

# Title
ax.text(
    5,
    13.7,
    "Methodology for Predicting 4-Year Median Earnings",
    ha="center",
    va="center",
    fontsize=16,
    fontweight="bold"
)

# Save diagram
plt.savefig(
    "methodology_diagram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
